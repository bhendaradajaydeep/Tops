##########################Topic 7#################

1. What is a dictionary?
   A dictionary in Python is a collection of data stored in key–value pairs. It is unordered, mutable (changeable), and indexed by keys.

2. What are key–value pairs?
   Key–value pairs are combinations where a unique key is used to store and access a specific value.
   Example: {"name": "Jay", "age": 25}

3. Can dictionary keys be duplicated?
   No, dictionary keys cannot be duplicated. If duplicate keys are used, the last value will overwrite the previous one.

4. What types are allowed as dictionary keys?
   Keys must be immutable (unchangeable) types such as:

* int
* float
* string
* tuple
  Mutable types like list, set, or dictionary cannot be keys.

5. Difference between dict["key"] and dict.get("key")?

* dict["key"]: Returns value but gives error if key is not found.
* dict.get("key"): Returns value but gives None (or default value) if key is not found.

6. How to update existing dictionary values?
   You can update values by assigning a new value to an existing key.
   Example:
   d["age"] = 30

7. What is a nested dictionary?
   A nested dictionary is a dictionary inside another dictionary.
   Example:
   {"student": {"name": "Jay", "age": 25}}

8. What is the use of .items()?
   .items() returns all key–value pairs as tuples. It is useful for looping through dictionary data.
   Example:
   for k, v in d.items():

9. What is a set?
   A set is an unordered collection of unique elements. It does not allow duplicate values.

10. Difference between list and set?
    List:

* Ordered
* Allows duplicates
* Uses []

Set:

* Unordered
* No duplicates allowed
* Uses {}

11. Why sets cannot contain duplicates?
    Sets are designed to store only unique values. Internally, they use hashing, which automatically removes duplicate entries.

12. Explain union, intersection, and difference with examples.
    Let A = {1, 2, 3} and B = {3, 4, 5}

* Union: All elements from both sets
  A ∪ B = {1, 2, 3, 4, 5}

* Intersection: Common elements
  A ∩ B = {3}

* Difference: Elements in A but not in B
  A − B = {1, 2}

13. Why are sets useful in deduplication?
    Sets automatically remove duplicate values, so they are useful when we want only unique data from a list or dataset.

14. Can a set contain a list? Why not?
    No, a set cannot contain a list because lists are mutable (changeable), and set elements must be immutable.

15. Where are sets used in data science?
    Sets are used in data science for:

* Removing duplicate data
* Finding common or unique values between datasets
* Data comparison
* Feature engineering
* Data cleaning processes

1. **What are conditional statements?**

12. **What is boolean expression?**
    An expression that returns True or False.

14. **How do you handle multiple conditions?**

################### Topic 11 ###################

1. What is a function in Python?
Answer:
A function in Python is a reusable block of code that performs a specific task. It helps avoid code repetition and makes programs easier to read, test, and maintain.

2. Difference between function and loop?
Answer:
Function:
- A function is used to organize and reuse code.
- It is called whenever needed.
- It can take input and return output.

Loop:
- A loop is used to repeat a block of code multiple times.
- It continues until a condition becomes false.
- It is mainly used for iteration.

3. What is return keyword?
Answer:
The return keyword is used to send a value back from a function to the place where the function was called. It also stops the execution of the function immediately.

4. Can a function return multiple values?
Answer:
Yes. A function can return multiple values by separating them with commas. Python automatically packs them into a tuple.

Example:
def values():
    return 10, 20, 30

5. What happens if a function has no return?
Answer:
If a function has no return statement, Python automatically returns None.

Example:
def greet():
    print("Hello")

print(greet())   # Output: None

6. Difference between parameter and argument?
Answer:
Parameter:
- A parameter is a variable defined in the function definition.

Argument:
- An argument is the actual value passed to the function when it is called.

Example:
def add(a, b):   # a and b are parameters
    return a + b

add(10, 20)      # 10 and 20 are arguments

7. What are default arguments?
Answer:
Default arguments are parameters that already have a default value. If no value is passed, the default value is used.

Example:
def greet(name="Guest"):
    print("Hello", name)

greet()      # Hello Guest
greet("John") # Hello John

8. What are keyword arguments?
Answer:
Keyword arguments are passed using the parameter name, making the code easier to read.

Example:
def student(name, age):
    print(name, age)

student(age=20, name="Rahul")

9. Can functions be nested?
Answer:
Yes. A function can be defined inside another function. This is called a nested function.

Example:
def outer():
    def inner():
        print("Inside Inner Function")
    inner()

10. What is a docstring?
Answer:
A docstring is a string written inside triple quotes just below the function definition. It explains what the function does.

Example:
def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

11. Why functions are important in analytics & data science?
Answer:
Functions are important because they:
- Reduce code repetition.
- Make code reusable.
- Improve readability.
- Simplify debugging.
- Help organize data analysis tasks.
- Make large projects easier to manage.

12. What happens after return is executed?
Answer:
After the return statement is executed:
- The function immediately stops running.
- The value is sent back to the caller.
- Any code written after return inside the function is not executed.

Example:
def test():
    return 10
    print("Hello")   # This line never runs

13. Can a function modify global variable?
Answer:
Yes. A function can modify a global variable using the global keyword.

Example:
count = 0

def update():
    global count
    count = count + 1

update()
print(count)   # Output: 1

14. What is the output type of multiple return values?
Answer:
The output type of multiple return values is a tuple because Python automatically packs multiple returned values into a tuple.

Example:
def values():
    return 1, 2, 3

result = values()
print(type(result))

1. What is a lambda function?

A lambda function is a small anonymous (nameless) function in Python that can have any number of arguments but only one expression. It is mainly used for short, simple operations.

Syntax:
lambda arguments: expression

Example:
square = lambda x: x * x
print(square(5))
# Output: 25


2. Difference between lambda and def?

Lambda:
- Anonymous function (no name required).
- Contains only one expression.
- Used for short and simple functions.
- Returns the result automatically.

def:
- Named function.
- Can contain multiple statements.
- Used for complex logic.
- Requires the return keyword to return a value.

Example:
def add(a, b):
    return a + b

add = lambda a, b: a + b


3. When should lambda be avoided?

Avoid lambda when:
- The function is long or complex.
- Multiple statements are needed.
- Loops or multiple conditions are required.
- Better readability is important.
- The function will be reused many times.


4. What does map() return?

map() returns a map object (an iterator). It does not return a list directly.

Example:
nums = [1, 2, 3]
result = map(lambda x: x * 2, nums)
print(list(result))
# Output: [2, 4, 6]


5. Difference between map() and for loop?

map():
- Applies a function to every element.
- Short and concise.
- Returns a map object.
- Best for simple transformations.

for loop:
- More flexible.
- Can perform multiple operations.
- Easier for complex logic.
- Returns nothing unless you store results.

Example:
Using map():
list(map(lambda x: x * 2, [1,2,3]))

Using for loop:
result = []
for x in [1,2,3]:
    result.append(x * 2)


6. What does filter() return?

filter() returns a filter object (an iterator) containing only the elements that satisfy the condition.

Example:
nums = [10, 15, 20, 25]
result = filter(lambda x: x > 15, nums)
print(list(result))
# Output: [20, 25]


7. Can filter() return an empty list?

Yes. If no elements satisfy the condition, filter() returns an empty iterator.

Example:
nums = [1, 2, 3]
result = filter(lambda x: x > 10, nums)
print(list(result))
# Output: []


8. What is reduce()?

reduce() repeatedly applies a function to combine all elements into a single value.

Example:
from functools import reduce

nums = [1,2,3,4]
result = reduce(lambda a, b: a + b, nums)
print(result)
# Output: 10


9. Why reduce() is in functools?

reduce() was moved to the functools module because it is a specialized functional programming tool and is not used as frequently as map() or filter().

Import:

10. Difference between sum() and reduce()?

sum():
- Adds numbers only.
- Built-in function.

reduce():
- Can perform any binary operation.

No. A lambda function can contain only one expression. For multiple statements, use a normal function with def.

1. What is recursion?
Recursion is a programming technique where a function calls itself to solve a problem. It breaks a large problem into smaller subproblems until a stopping condition (base case) is reached.

2. What is base case?
A base case is the condition that stops a recursive function from calling itself. It prevents infinite recursion.

3. What happens if base case is missing?
If the base case is missing, the function keeps calling itself forever until Python raises a RecursionError (stack overflow).

4. What is call stack?
The call stack is a memory structure that stores information about active function calls. Each function call is added to the stack and removed when it finishes.

5. Difference between recursion and loop?
Recursion:
- A function calls itself.
- Uses the call stack.
- Easier for problems like trees and graphs.

Loop:
- Uses for or while.
- Does not use the call stack for repetition.
- Usually faster and uses less memory.

6. What is stack overflow?
Stack overflow occurs when too many recursive function calls fill the call stack, causing the program to crash with a RecursionError.

7. When should recursion be avoided?
Recursion should be avoided when:
- The problem can be solved easily with loops.
- Performance and memory are important.
- Deep recursion may cause a stack overflow.

8. What is variable scope?
Variable scope is the region of a program where a variable can be accessed.

Local Variable:
- Declared inside a function.

Global Variable:
- Declared outside all functions.
- Accessible throughout the program.

11. What is nonlocal keyword?

12. Can Python functions access outer variables?
Yes. Python functions can access variables from their outer (enclosing) scope. This is called a closure or enclosing scope.

13. What is built-in scope?
Built-in scope contains Python's predefined functions and objects such as print(), len(), input(), max(), min(), and sum().

14. Why overwriting built-ins is dangerous?
Overwriting built-in names (like list, sum, or max) replaces the original built-in function with your variable, which can cause errors and unexpected behavior in your program.
