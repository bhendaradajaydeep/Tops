```python
import pandas as pd

df = pd.DataFrame({
    'username': [
        'CoolDude',
        'foodieQueen',
        'TECH_GUY',
        'TravelLover',
        'MusicFan'
    ]
})

df['username'] = df['username'].str.lower()

print(df)
```

```python
import pandas as pd

songs = pd.Series([
    'Shape of You Remix',
    'Blinding Lights Remix',
    'Perfect',
    'Kesariya Remix',
    'Dance Remix Party'
])

cleaned_songs = songs.str.replace('Remix', '', regex=False).str.strip()

print(cleaned_songs)
```

```python
import pandas as pd

df = pd.DataFrame({
    'review': [
        'Fast delivery and good food',
        'The food was tasty',
        'Delivery was very late',
        'Excellent service',
        'Quick DELIVERY, nice packaging'
    ]
})

result = df[
    df['review'].str.contains('delivery', case=False, na=False)
]

print(result)
```

```python
import pandas as pd

categories = pd.Series([
    'Electronics/Mobiles/Smartphones',
    'Home/Kitchen/Appliances',
    'Fashion/Men/Shirts',
    'Beauty/Skincare/Creams',
    'Sports/Fitness/Equipment'
])

df = pd.DataFrame({
    'CategoryPath': categories
})

df['MainCategory'] = df['CategoryPath'].str.split('/').str[0]

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'title': [
        'Python for Beginners',
        'Power BI Tutorial',
        'Excel Data Analysis',
        'Learn Pandas',
        'YouTube Growth Tips'
    ]
})

# Transform in a single line
df['title'] = df['title'].str.upper().str.replace(' ', '_', regex=False)

print(df)
```
