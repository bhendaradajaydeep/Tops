```python
import pandas as pd

# Create sample product sales data
data = {
    'ProductName': ['Laptop', 'Mobile', 'Headphones', 'Keyboard', 'Mouse'],
    'Price': [50000, 20000, 1500, 1200, 500],
    'Qty': [2, 3, 4, 2, 5]
}

df = pd.DataFrame(data)

# Save data as CSV
df.to_csv('flipkart_product_sales.csv', index=False)

# Load CSV file
df = pd.read_csv('flipkart_product_sales.csv')

# Add TotalValue column
df['TotalValue'] = df['Price'] * df['Qty']

print(df)
```

```python
import pandas as pd

data = {
    'Item': ['Pizza', 'Burger', 'Sandwich', 'Pasta', 'Fries'],
    'Price': [300, 150, 120, 250, 100],
    'Qty': [2, 3, 1, 2, 4]
}

df = pd.DataFrame(data)

# Apply 10% discount
df['DiscountedPrice'] = df['Price'].apply(lambda price: price * 0.90)

print(df)
```

```python
import pandas as pd

data = {
    'PostID': [101, 102, 103, 104],
    'Likes': [1500, 2000000, 850, 1250000],
    'Comments': [250, 1200, 2000000, 45000]
}

df = pd.DataFrame(data)

# Helper function to format large numbers
def format_number(num):
    if num >= 1_000_000:
        return f"{num / 1_000_000:g}M"
    elif num >= 1_000:
        return f"{num / 1_000:g}K"
    else:
        return str(num)

# Apply formatting to Likes and Comments
columns = ['Likes', 'Comments']

# Modern pandas method
df[columns] = df[columns].map(format_number)

print(df)
```

```python
import pandas as pd

data = {
    'Restaurant': ['Dominos', 'Pizza Hut', 'KFC', 'Subway', 'Burger King'],
    'Price': [300, 400, 250, 180, 200],
    'Qty': [2, 1, 3, 2, 2],
    'DeliveryCharge': [30, 40, 25, 20, 30]
}

df = pd.DataFrame(data)

# Calculate final order amount
df['FinalAmount'] = df.apply(
    lambda row: (row['Price'] * row['Qty']) + row['DeliveryCharge'],
    axis=1
)

print(df)
```

```python
import pandas as pd

data = {
    'Product': ['T-Shirt', 'Jeans', 'Shoes', 'Jacket', 'Shirt'],
    'Price': [800, 1500, 2500, 3000, 1000],
    'Qty': [2, 1, 1, 2, 3]
}

df = pd.DataFrame(data)

# Calculate GST at 5% of the total product value
df['GST'] = df['Price'] * df['Qty'] * 0.05

print(df)
```
