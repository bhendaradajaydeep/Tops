```python
import pandas as pd

dates = pd.Series([
    '15-01-2026',
    '23-03-2026',
    '10-05-2026',
    '28-07-2026',
    '05-10-2026'
])

# Convert strings into datetime objects
dates = pd.to_datetime(dates, format='%d-%m-%Y')

print(dates)
print("\nData Type:", dates.dtype)
```

```python
import pandas as pd

df = pd.DataFrame({
    'movie': [
        'Pathaan',
        'Jawan',
        'Dangal',
        'Animal',
        '3 Idiots'
    ],
    'release_date': [
        '25-01-2023',
        '07-09-2023',
        '23-12-2016',
        '01-12-2023',
        '25-12-2009'
    ]
})

# Convert release_date into datetime
df['release_date'] = pd.to_datetime(
    df['release_date'],
    format='%d-%m-%Y'
)

# Extract year, month and day
df['year'] = df['release_date'].dt.year
df['month'] = df['release_date'].dt.month
df['day'] = df['release_date'].dt.day

print(df)
```

```python
import pandas as pd

# Generate dates starting today, including today
dates = pd.date_range(
    start=pd.Timestamp.today().normalize(),
    periods=7,
    freq='D'
)

print("Next 7 Days:")
print(dates)
```

```python
import pandas as pd

df = pd.DataFrame({
    'user_id': [101, 102, 103, 104, 105],
    'start_date': [
        '01-01-2026',
        '15-03-2026',
        '01-06-2026',
        '20-08-2026',
        '01-10-2026'
    ]
})

# Convert start_date to datetime
df['start_date'] = pd.to_datetime(
    df['start_date'],
    format='%d-%m-%Y'
)

# Get today's date
today = pd.Timestamp.today().normalize()

# Calculate tenure in days
df['tenure_days'] = (today - df['start_date']).dt.days

print(df)
```
