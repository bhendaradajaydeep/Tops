```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("IPL_2023_Batsmen.csv")

# Display first 10 rows
print(df.head(10))
```

```python
# Convert runs to numeric values if necessary
df['runs'] = pd.to_numeric(df['runs'], errors='coerce')

print("Mean:", df['runs'].mean())
print("Median:", df['runs'].median())
print("Mode:", df['runs'].mode().tolist())
print("Minimum:", df['runs'].min())
print("Maximum:", df['runs'].max())
```

```python
import pandas as pd

restaurants = pd.DataFrame({
    'name': [
        'Restaurant A',
        'Restaurant B',
        'Restaurant C',                
        'Restaurant D',
        'Restaurant E'
    ],
    'rating': [4.5, 4.2, 3.8, 4.7, 4.0],
    'votes': [1200, 900, 400, 1500, 650],
    'avg_cost': [500, 350, 250, 700, 300]
})

print(restaurants)

# Correlation between rating and votes
correlation = restaurants['rating'].corr(restaurants['votes'])

print("\nCorrelation coefficient:", round(correlation, 3))
```
