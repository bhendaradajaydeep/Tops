```python
import pandas as pd

ipl = pd.DataFrame({
    'team': [
        'CSK', 'MI', 'RCB', 'CSK',
        'MI', 'RCB', 'CSK', 'MI'
    ],
    'runs': [180, 160, 200, 150, 175, 190, 210, 145]
})

# Total runs by team
team_runs = ipl.groupby('team')['runs'].sum().reset_index()

print(team_runs)
```

```python
import pandas as pd

orders = pd.DataFrame({
    'city': ['Ahmedabad', 'Ahmedabad', 'Ahmedabad',
             'Surat', 'Surat', 'Surat'],
    'restaurant': ['Pizza Hub', 'Pizza Hub', 'Burger Point',
                   'Pizza Hub', 'Burger Point', 'Burger Point'],
    'order_amount': [200, 300, 150, 400, 250, 350]
})

result = orders.groupby(
    ['city', 'restaurant']
).agg(
    average_order_amount=('order_amount', 'mean')
).reset_index()

print(result)
```

```python
products = pd.DataFrame({
    'category': ['Electronics', 'Electronics', 'Fashion',
                 'Fashion', 'Electronics', 'Fashion'],
    'region': ['West', 'West', 'North', 'North', 'South', 'West'],
    'sales': [50000, 30000, 20000, 25000, 40000, 15000],
    'units_sold': [50, 30, 40, 50, 20, 30]
})

result = products.groupby(
    ['category', 'region']
).agg(
    total_sales=('sales', 'sum'),
    average_units_sold=('units_sold', 'mean')
).reset_index()

print(result)
```

```python
revenue = pd.DataFrame({
    'region': ['West', 'West', 'North', 'North', 'South', 'South'],
    'revenue': [50000, 70000, 30000, 45000, 60000, 40000]
})

result = revenue.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
})

print(result)
```

```python
songs = pd.DataFrame({
    'artist': ['Arijit', 'Neha', 'Arijit', 'Badshah',
               'Neha', 'Badshah', 'Arijit'],
    'genre': ['Pop', 'Pop', 'Rock', 'Hip-Hop',
              'Rock', 'Hip-Hop', 'Pop'],
    'streams': [1000, 2000, 1500, 3000, 2500, 2000, 3000]
})

result = songs.groupby('genre').agg(
    total_streams=('streams', 'sum'),
    average_streams=('streams', 'mean')
)

result = result.sort_values(
    by='total_streams',
    ascending=False
)

print(result)
```
