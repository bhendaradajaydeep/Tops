```python
import pandas as pd

movies = pd.DataFrame({
    'title': [
        '3 Idiots', 'Dangal', 'Pathaan',
        'Drishyam 2', 'PK', 'Sholay'
    ],
    'rating': [8.4, 8.3, 5.8, 8.2, 8.1, 8.1],
    'release_year': [2009, 2016, 2023, 2022, 2014, 1975]
})

# Sort movies by rating from highest to lowest
result = movies.sort_values(by='rating', ascending=False)

print(result)
```

```python
import pandas as pd

ipl = pd.DataFrame({
    'team': ['CSK', 'MI', 'RCB', 'KKR', 'RR'],
    'points': [17, 16, 14, 20, 12],
    'net_run_rate': [0.65, 0.42, 0.18, 1.10, -0.25]
})

# Sort index in descending order
result = ipl.sort_index(ascending=False)

print(result)
```

```python
import pandas as pd

youtube = pd.DataFrame({
    'video_title': [
        'Cricket Highlights',
        'Bollywood Trailer',
        'Music Video',
        'Tech Review'
    ],
    'views': [5000000, 3200000, 7000000, 1500000],
    'likes': [250000, 180000, 400000, 75000]
}, index=[10, 20, 30, 40])

# Reset index without keeping the old index
result = youtube.reset_index(drop=True)

print(result)
```

```python
import pandas as pd

restaurants = pd.DataFrame({
    'restaurant': [
        'Pizza Hut', 'McDonalds', 'Dominos',
        'Burger King', 'Subway'
    ],
    'rating': [4.1, 4.0, 4.3, 4.2, 3.9]
})

# Rearrange row indexes to put Dominos first
new_order = [2, 0, 1, 3, 4]

result = restaurants.reindex(new_order).reset_index(drop=True)

print(result)
```

```python
import pandas as pd

products = pd.DataFrame({
    'product_name': [
        'Laptop', 'Mobile', 'Headphones',
        'Smartwatch', 'Keyboard', 'Monitor'
    ],
    'price': [55000, 25000, 2000, 4000, 1500, 12000],
    'discount': [10, 25, 40, 30, 15, 20]
})

# Sort by discount in descending order and reset index
result = products.sort_values(
    by='discount',
    ascending=False
).reset_index(drop=True)

print(result)
```
