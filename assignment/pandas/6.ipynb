```python
import pandas as pd

data = {
    'product_id': [101, 102, 101, 103, 102, 101, 104],
    'user_id':    [1,   2,   1,   3,   2,   4,   5],
    'rating':     [5,   4,   3,   5,   4,   4,   3],
    'review_text': [
        'Excellent', 'Good', 'Average',
        'Very good', 'Nice', 'Worth buying', 'Okay'
    ]
}

df = pd.DataFrame(data)

# Count duplicate rows before cleaning
duplicates_before = df.duplicated(
    subset=['product_id', 'user_id']
).sum()

print("Duplicate reviews before cleaning:", duplicates_before)

# Remove duplicates, keeping the first review
df_cleaned = df.drop_duplicates(
    subset=['product_id', 'user_id'],
    keep='first'
)

# Count duplicates after cleaning
duplicates_after = df_cleaned.duplicated(
    subset=['product_id', 'user_id']
).sum()

print("Duplicate reviews after cleaning:", duplicates_after)

print("\nCleaned DataFrame:")
print(df_cleaned)
```

```python
import pandas as pd

data = {
    'order_id': [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],
    'order_amount': [
        250, 300, 350, 280, 400,
        320, 270, 310, 290, 5000
    ]
}

df = pd.DataFrame(data)

# Calculate Q1 and Q3
Q1 = df['order_amount'].quantile(0.25)
Q3 = df['order_amount'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Find outliers
outliers = df[
    (df['order_amount'] < lower_limit) |
    (df['order_amount'] > upper_limit)
]

print("\nOutlier Orders:")
print(outliers[['order_id', 'order_amount']])
```

```python
import pandas as pd
from scipy.stats import zscore

data = {
    'date': [
        '2026-04-01', '2026-04-02', '2026-04-03',
        '2026-04-04', '2026-04-05', '2026-04-06',
        '2026-04-07', '2026-04-08', '2026-04-09',
        '2026-04-10'
    ],
    'tickets_sold': [
        1200, 1350, 1100, 1400, 1250,
        1300, 1150, 1280, 1220, 10000
    ]
}

df = pd.DataFrame(data)

# Calculate Z-score
df['z_score'] = zscore(df['tickets_sold'])

print("Data with Z-scores:")
print(df)

# Find dates where absolute Z-score is greater than 2
outliers = df[df['z_score'].abs() > 2]

print("\nOutlier Dates:")
print(outliers[['date', 'tickets_sold', 'z_score']])
```

```python
import pandas as pd
from scipy.stats import zscore

# Function to find transaction outliers
def find_outliers(df, column='transaction_amount', threshold=2):
    scores = zscore(df[column])

    # Return transactions with absolute Z-score above threshold
    outliers = df[scores.abs() > threshold]

    return outliers


# Sample Paytm wallet transactions
data = {
    'transaction_id': [501, 502, 503, 504, 505,
                       506, 507, 508, 509, 510],
    'user_name': [
        'Aarav', 'Diya', 'Krish', 'Meera', 'Ishaan',
        'Anaya', 'Vivaan', 'Riya', 'Arjun', 'Kavya'
    ],
    'transaction_amount': [
        200, 250, 300, 150, 280,
        220, 350, 270, 240, 10000
    ]
}

df = pd.DataFrame(data)

# Test the function
result = find_outliers(df)

print("Outlier Transactions:")
print(result)
```
