```python
import pandas as pd

followers = pd.Series([1200, 54000, 780, 150000, 32000])

highest_index = followers.idxmax()
lowest_index = followers.idxmin()

print("Highest Followers:")
print("Influencer", highest_index, "=", followers[highest_index])

print("\nLowest Followers:")
print("Influencer", lowest_index, "=", followers[lowest_index])
```
