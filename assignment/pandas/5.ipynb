```python
import pandas as pd

df = pd.read_csv("ipl_players.csv")

# Display rows where age is missing
result = df[df["age"].isna()]

print(result)
```

```python
import pandas as pd

df = pd.read_csv("food_delivery_orders.csv")

# Remove rows with missing delivery ratings
cleaned_df = df.dropna(subset=["delivery_rating"])

print(cleaned_df)
```

```python
import pandas as pd

df = pd.read_csv("flipkart_product_reviews.csv")

# Calculate mean rating
mean_rating = df["rating"].mean()

# Fill missing ratings with the mean
df["rating"] = df["rating"].fillna(mean_rating)

print(df)
```

```python
import pandas as pd

df = pd.read_csv("spotify_playlist.csv")

# Calculate median duration
median_duration = df["duration"].median()

# Fill missing durations with median
df["duration"] = df["duration"].fillna(median_duration)

# Save cleaned data to a new CSV file
df.to_csv("spotify_playlist_cleaned.csv", index=False)

print(df)
print("Cleaned CSV file saved successfully!")
```

```python
import pandas as pd

df = pd.read_csv("instagram_users.csv")

# Replace missing followers with 0
df["followers"] = df["followers"].fillna(0)

print(df)

# Save updated data
df.to_csv("instagram_users_cleaned.csv", index=False)
```
