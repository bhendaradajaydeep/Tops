```python
import pandas as pd

# Users DataFrame
users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'city': ['Ahmedabad', 'Junagadh', 'Rajkot', 'Surat', 'Baroda']
})

# Orders DataFrame
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 5, 2, 7],
    'order_amount': [500, 800, 300, 1200, 450, 600, 900]
})

# Inner Join
result = pd.merge(users, orders, on='user_id', how='inner')

print(result)
```

```python
result = pd.merge(users, orders, on='user_id', how='left')

print(result)
```

```python
restaurants = pd.DataFrame({
    'restaurant_id': [1, 2, 3, 4],
    'name': ['Pizza Hub', 'Burger Point', 'Spice House', 'Food Corner'],
    'city': ['Ahmedabad', 'Rajkot', 'Surat', 'Junagadh']
})

zomato_reviews = pd.DataFrame({
    'review_id': [101, 102, 103, 104],
    'restaurant_id': [1, 2, 2, 5],
    'rating': [4.5, 4.0, 3.5, 5.0],
    'reviewer_city': ['Ahmedabad', 'Rajkot', 'Surat', 'Mumbai']
})

result = pd.merge(
    restaurants,
    zomato_reviews,
    on='restaurant_id',
    how='outer'
)

print(result)
```

```python
playlists = pd.DataFrame({
    'playlist_id': [101, 102, 103, 104],
    'user_id': [1, 2, 3, 4],
    'city': ['Ahmedabad', 'Rajkot', 'Surat', 'Junagadh'],
    'genre': ['Pop', 'Rock', 'Jazz', 'Hip-Hop']
})

user_profiles = pd.DataFrame({
    'user_id': [1, 2, 3, 4],
    'username': ['Amit', 'Priya', 'Rahul', 'Neha'],
    'city': ['Ahmedabad', 'Surat', 'Surat', 'Junagadh']
})

result = pd.merge(
    playlists,
    user_profiles,
    on=['user_id', 'city'],
    how='inner'
)

print(result)
```

```python
products = pd.DataFrame({
    'product_id': [1, 2, 3, 4],
    'name': ['Laptop', 'Mobile', 'Headphones', 'Keyboard'],
    'category': ['Electronics', 'Electronics', 'Accessories', 'Accessories']
})

ratings = pd.DataFrame({
    'rating_id': [101, 102, 103, 104, 105],
    'product_id': [1, 2, 2, 5, 3],
    'rating': [4.5, 4.0, 3.5, 5.0, 4.2]
})

# Right Join
result = pd.merge(
    products,
    ratings,
    on='product_id',
    how='right'
)

print(result)
```
