```python
import pandas as pd

data = {
    'restaurant': [
        'Dominos', 'Pizza Hut', 'Burger King',
        'McDonalds', 'Subway', 'Haldirams',
        'KFC', 'La Pinoz'
    ],
    'order_amount': [
        350, 750, 450, 620, 280, 900, 550, 1200
    ],
    'delivery_time': [
        30, 40, 25, 35, 20, 45, 32, 50
    ]
}

df = pd.DataFrame(data)

# Filter orders greater than 500
result = df[df['order_amount'] > 500]

print(result)
```

```python
import pandas as pd

data = {
    'song': [
        'Shape of You', 'Blinding Lights',
        'Perfect', 'Levitating', 'Believer',
        'Flowers', 'Stay', 'Attention'
    ],
    'artist': [
        'Ed Sheeran', 'The Weeknd',
        'Ed Sheeran', 'Dua Lipa', 'Imagine Dragons',
        'Miley Cyrus', 'Justin Bieber', 'Charlie Puth'
    ],
    'streams': [
        2500000, 3200000, 900000, 1800000,
        2100000, 1500000, 800000, 1200000
    ],
    'duration': [
        240, 200, 263, 175, 204, 170, 141, 210
    ]
}

df = pd.DataFrame(data)

# Filter using query()
result = df.query('streams > 1000000 and duration < 180')

print(result)
```

```python
import pandas as pd

data = {
    'team1': [
        'CSK', 'MI', 'RCB', 'KKR', 'RR',
        'GT', 'DC', 'SRH', 'PBKS', 'LSG'
    ],
    'team2': [
        'MI', 'CSK', 'KKR', 'RCB', 'GT',
        'RR', 'SRH', 'DC', 'LSG', 'PBKS'
    ],
    'winner': [
        'CSK', 'MI', 'RCB', 'KKR', 'RR',
        'GT', 'SRH', 'SRH', 'PBKS', 'LSG'
    ],
    'total_runs': [
        175, 195, 220, 185, 230,
        205, 160, 215, 180, 240
    ]
}

df = pd.DataFrame(data)

# Boolean indexing: between 180 and 220, inclusive
result = df[
    (df['total_runs'] >= 180) &
    (df['total_runs'] <= 220)
]

print(result)
```
