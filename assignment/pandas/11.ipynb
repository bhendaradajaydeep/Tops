```python
import pandas as pd

# January orders
jan_orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'customer': ['Amit', 'Priya', 'Rahul'],
    'food_item': ['Pizza', 'Burger', 'Pasta'],
    'amount': [250, 150, 300]
})

# February orders
feb_orders = pd.DataFrame({
    'order_id': [104, 105, 106],
    'customer': ['Neha', 'Karan', 'Riya'],
    'food_item': ['Sandwich', 'Dosa', 'Pizza'],
    'amount': [120, 100, 280]
})

# Combine row-wise
all_orders = pd.concat(
    [jan_orders, feb_orders],
    ignore_index=True
)

print(all_orders)
```

```python
products = pd.DataFrame({
    'product_id': [1, 2, 3, 4],
    'product_name': ['Laptop', 'Mobile', 'Headphones', 'Keyboard']
})

ratings = pd.DataFrame({
    'rating': [4.5, 4.2, 3.9, 4.7]
})

# Combine column-wise
product_ratings = pd.concat(
    [products, ratings],
    axis=1
)

print(product_ratings)
```

```python
streams = pd.DataFrame({
    'user': ['Amit', 'Amit', 'Amit', 'Priya', 'Priya', 'Priya',
             'Rahul', 'Rahul', 'Rahul'],
    'day': ['Monday', 'Tuesday', 'Wednesday',
            'Monday', 'Tuesday', 'Wednesday',
            'Monday', 'Tuesday', 'Wednesday'],
    'streams': [120, 150, 180, 200, 220, 210, 90, 110, 130]
})

# Reshape the data
pivot_data = streams.pivot(
    index='user',
    columns='day',
    values='streams'
)

print(pivot_data)
```

```python
orders = pd.DataFrame({
    'restaurant': ['Pizza Hub', 'Pizza Hub', 'Pizza Hub',
                   'Burger Point', 'Burger Point', 'Burger Point'],
    'month': ['January', 'January', 'February',
              'January', 'February', 'February'],
    'total_amount': [200, 300, 400, 150, 250, 350]
})

# Calculate average order amount
monthly_average = orders.pivot_table(
    index='restaurant',
    columns='month',
    values='total_amount',
    aggfunc='mean'
)

print(monthly_average)
```

```python
import pandas as pd

# Create January sales data
jan = pd.DataFrame({
    'product': ['T-Shirt', 'Jeans', 'Shoes'],
    'sales': [10000, 15000, 20000]
})
jan.to_csv('sales_jan.csv', index=False)

# Create February sales data
feb = pd.DataFrame({
    'product': ['T-Shirt', 'Jeans', 'Shoes'],
    'sales': [12000, 18000, 22000]
})
feb.to_csv('sales_feb.csv', index=False)

# Create March sales data
mar = pd.DataFrame({
    'product': ['T-Shirt', 'Jeans', 'Shoes'],
    'sales': [15000, 17000, 25000]
})
mar.to_csv('sales_mar.csv', index=False)

# Read all three CSV files
jan = pd.read_csv('sales_jan.csv')
feb = pd.read_csv('sales_feb.csv')
mar = pd.read_csv('sales_mar.csv')

# Combine all monthly data
all_sales = pd.concat(
    [jan, feb, mar],
    ignore_index=True
)

print("Combined Sales Data:")
print(all_sales)

# Calculate total sales per product
total_sales = all_sales.groupby('product')['sales'].sum()

print("\nTotal Sales Per Product:")
print(total_sales)
```
