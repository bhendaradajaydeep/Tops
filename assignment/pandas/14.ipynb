```python
import pandas as pd
import numpy as np

# Load the dataset
ipl_df = pd.read_csv("IPL_Match.csv")

# Display first 5 rows
print(ipl_df.head())

# Check column names
print(ipl_df.columns.tolist())
```

```python
conditions = [
    ipl_df['win_by_runs'] > 0,
    ipl_df['win_by_wickets'] > 0
]

choices = ['Runs', 'Wickets']

ipl_df['match_margin_type'] = np.select(
    conditions,
    choices,
    default='Tie/No Result'
)

print(ipl_df[
    ['winner', 'win_by_runs', 'win_by_wickets', 'match_margin_type']
].head(10))
```

```python
team_cities = pd.DataFrame({
    'team': [
        'Chennai Super Kings',
        'Mumbai Indians',
        'Royal Challengers Bangalore',
        'Kolkata Knight Riders',
        'Rajasthan Royals',
        'Delhi Capitals',
        'Punjab Kings',
        'Sunrisers Hyderabad',
        'Gujarat Titans',
        'Lucknow Super Giants'
    ],
    'home_city': [
        'Chennai',
        'Mumbai',
        'Bengaluru',
        'Kolkata',
        'Jaipur',
        'Delhi',
        'Mohali',
        'Hyderabad',
        'Ahmedabad',
        'Lucknow'
    ]
})

# Add home city for team1
ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={
        'team': 'team1',
        'home_city': 'team1_home_city'
    }),
    on='team1',
    how='left'
)

# Add home city for team2
ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={
        'team': 'team2',
        'home_city': 'team2_home_city'
    }),
    on='team2',
    how='left'
)

print(ipl_df[
    ['team1', 'team1_home_city', 'team2', 'team2_home_city']
].head(10))
```

```python
# Add a count column for pivot table aggregation
ipl_df['match_count'] = 1

# Pivot table: team wins by season
wins_pivot = ipl_df.pivot_table(
    index='winner',
    columns='season',
    values='match_count',
    aggfunc='sum',
    fill_value=0
)

print("Matches won by each team per season:")
print(wins_pivot)

# Find the top 3 team-season combinations
top_3_wins = (
    ipl_df.groupby(['season', 'winner'])
    .size()
    .reset_index(name='total_wins')
    .sort_values('total_wins', ascending=False)
    .head(3)
)

print("\nTop 3 team-season combinations:")
print(top_3_wins)
```
