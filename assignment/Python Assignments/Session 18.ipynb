```python
# Use re.findall() to extract all valid Indian phone numbers (10 digits, starting with 7, 8, or 9) from a given text string that contains 
# random numbers, prices, and phone numbers like those seen in OLX or WhatsApp chats.

import re

text = """
Call me at 9876543210 or 8123456789.
My old number was 6123456789 (invalid).
Price is 5000 or 9999.
Another number: 7894561230.
Fake: 1234567890
WhatsApp: 9988776655
"""
phones=re.findall(r"[678]\d{9}",text)

print(phones)
```

```python
import re

def check_order_id(text):
    pattern = r"\bOD\d{18}\b"
    
    if re.search(pattern, text):
        return True
    else:
        return False


samples = [
    "Your order OD123456789012345000 has been shipped",
    "Invalid order OD12345",
    "Check this ID: OD987654321098765432",
    "Random text without ID",
    "Another one: OD111122223333444455"
]

for s in samples:
    print(s, "->", check_order_id(s))
```
