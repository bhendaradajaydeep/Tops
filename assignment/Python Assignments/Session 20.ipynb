```python
# Create a Product class with a private attribute _price

class Product:
    def __init__(self, price):
        self._price = price

    def display_price(self):
        print("Price:", self._price)

# Create object
product = Product(500)

# Display price
product.display_price()
```

```python
# Add getter and setter methods for _price

class Product:
    def __init__(self, price):
        self._price = price

    # Getter method
    def get_price(self):
        return self._price

    # Setter method
    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative.")
        self._price = new_price

# Create object
product = Product(500)

# Get price
print("Current Price:", product.get_price())

# Update price
product.set_price(750)
print("Updated Price:", product.get_price())

# Try setting a negative price
# product.set_price(-100)   # This will raise ValueError
```

```python
# Create Playlist class with private attribute _songs

class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
        else:
            print("Song not found.")

    def get_songs(self):
        return self._songs

# Create object
playlist = Playlist()

# Add songs
playlist.add_song("Blinding Lights")
playlist.add_song("Shape of You")

# Remove a song
playlist.remove_song("Shape of You")

# Display songs
print("Playlist:", playlist.get_songs())
```

```python
# Create an abstract class PaymentMethod

from abc import ABC, abstractmethod

class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

# UPI subclass
class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI.")

# CreditCard subclass
class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card.")

# Create objects
upi = UPI()
card = CreditCard()

# Make payments
upi.pay(500)
card.pay(1200)
```
